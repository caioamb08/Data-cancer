```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.pipeline import Pipeline
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
df = pd.read_csv('data_cancer2.csv')
```

```python
df
```

```python
df.shape
```

```python
#começar a limpeza
df = df.drop(columns=["Unnamed: 32"])
```

```python
df.diagnosis = df.diagnosis.map({"M": 1, "B": 0})
```

```python
print(df["diagnosis"].value_counts())  
print(df.dtypes.value_counts())      
print(df.isna().sum().sum())
```

```python
#diagnosis é o ponto alvo que preciso  entao sera o y
X = df.drop(columns=['diagnosis','id'])
y = df.diagnosis
```

```python
print(X.shape)           
print(y.shape)   
```

```python
#df tratado, agora posso começar meu modelo de ml
```

```python
X_train,X_test,y_train,y_test = train_test_split(X, y, test_size = 0.3, random_state = 0, stratify= y)
```

```python
print(X_train.shape, X_test.shape)
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import StandardScaler
```

```python
#começar o modelo pelo naive, testar todos algoritimos
# colocar os numeros em escala, pois tem problemas de overfitting
pipe_nb = Pipeline([
    ('scaler',StandardScaler()),
    ('modelo',GaussianNB()),
])
```

```python
#treinar o algoritimo
pipe_nb.fit(X_train,y_train)
```

```python
#fazer a previsão e analisar
pred_nb = pipe_nb.predict(X_test)
pred_nb
```

```python
#ver o resultado do teste
print("Acurácia:", accuracy_score(y_test, pred_nb)*100)
print(confusion_matrix(y_test, pred_nb))
print(classification_report(y_test, pred_nb))
```

```python
#validação cruada
from sklearn.model_selection import StratifiedKFold, cross_val_score

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)

scores = cross_val_score(pipe_nb, X_train, y_train, cv=cv, scoring="accuracy")
          
print(scores.mean()*100)    
```

```python
#valor do treino
print(pipe_nb.score(X_train, y_train)*100)
```

```python
#pipeline com svm
from sklearn.svm import SVC

pipe_svm = Pipeline([
    ("scaler", StandardScaler()),
    ("modelo", SVC()),
])
```

```python
#separar x e y
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0, stratify=y)
```

```python
#validação cruzada com 5 splits
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)
```

```python
#treino
pipe_svm.fit(X_train, y_train)
print("Treino:", pipe_svm.score(X_train, y_train))
```

```python
#validação só com o treino
scores_svm = cross_val_score(pipe_svm, X_train, y_train, cv=cv, scoring="accuracy")
print("Validação cruzada:", scores_svm.mean())
```

```python
#teste
pred_svm = pipe_svm.predict(X_test)
print("Teste:", accuracy_score(y_test, pred_svm))
print(confusion_matrix(y_test, pred_svm))
print(classification_report(y_test, pred_svm))
```
